````python
# DTSC 520 setup. Run this once, first, every session. Sets up %%expect.
import builtins, sys, traceback
from IPython import get_ipython
from IPython.core.magic import register_cell_magic

@register_cell_magic
def expect(name, cell):
    """%%expect ErrorName -- run a cell that is supposed to fail."""
    want, rule = name.strip() or "Exception", "─" * 64
    try:
        code = compile(cell, "<expected to fail>", "exec")
    except SyntaxError as e:
        raised, tb = e, None
    else:
        try:
            exec(code, get_ipython().user_ns)
            print(f"{rule}\n  NO ERROR RAISED. We expected a {want} here.\n{rule}")
            return
        except BaseException as e:
            raised, tb = e, e.__traceback__.tb_next
    print(f"{rule}\n  EXPECTED ERROR. This cell is supposed to fail.\n{rule}")
    traceback.print_exception(type(raised), raised, tb, limit=4, file=sys.stdout)
    if not isinstance(raised, getattr(builtins, want, ())):
        print(f"\n  (note: expected {want}, got {type(raised).__name__})")

print("Setup complete. %%expect is ready.")
````

## The vocabulary

matplotlib's vocabulary is small and its error
messages assume you know it. The first four entries in particular are worth
learning early, because almost every question you ask online will be answered in
those words.

- **Figure** - The whole image. It owns the figure-level title, the size, and saving to a file.
- **Axes** - One plot inside a Figure. It owns the data, the labels, the legend, and the two axis objects. An `Axes` is not an axis.
- **Axis** - A single scale on an Axes, with its ticks, limits and formatting.
- **Artist** - Anything matplotlib draws. Every line, dot, label and spine is one.
- **explicit interface** - Creating `fig, ax` and calling methods on them. Also called the object-oriented interface, and the only one this course uses.
- **implicit interface** - The `plt.plot` style, which adds to whatever Axes matplotlib thinks is current. Convenient for one throwaway chart, and the reason a second chart lands in the wrong place.
- **constrained layout** - The layout engine requested with `layout="constrained"`, which sizes the gaps so labels do not collide. It replaces `tight_layout`, and calling `tight_layout()` switches it off.
- **bin** - One interval a histogram counts values into. The bin count is a choice you make, and it changes the shape the reader sees.
- **alpha** - Opacity, from 0 to 1. Below 1, overlapping points look darker than single ones, which is how a scatter plot admits that two students landed in the same place.
- **colormap** - The mapping from a number to a color. Sequential for ordered values, diverging for a meaningful middle, qualitative for unordered categories.
- **perceptually uniform** - A colormap in which equal steps in the data look like equal steps in color. `viridis` is; `jet` is not, which is why it invents boundaries the data does not have.
- **colorbar** - The key for a colormap. Ask the Figure for it: `fig.colorbar(mappable, ax=ax)`.
- **relative luminance** - How bright a color is, and therefore what gray it prints as. Two colors of similar luminance are the same color to a monochrome printer.
- **small multiples** - A grid of panels sharing a scale, so several comparisons can be read at once. `sharey=True` is what makes them comparable.
- **dpi** - Dots per inch when a figure is saved as a raster image. 72 is a thumbnail, 150 a document figure, 300 print.
- **raster and vector** - Raster formats (PNG) store pixels and blur when enlarged. Vector formats (PDF, SVG) store drawing instructions and do not.
- **axes-level function** - A seaborn function that draws onto an Axes you pass with `ax=`, such as `histplot` or `regplot`.
- **figure-level function** - A seaborn function that builds its own figure and cannot take `ax=`, such as `pairplot` or `displot`. Customize it through the grid object it returns.
- **truncated axis** - A value axis that does not start at zero. On a bar chart, where the reader compares lengths, it misrepresents the comparison.
- **confounder** - A third variable that produces an apparent relationship between two others. Faceting or coloring by it is the visual control.

# 1. Why draw it at all

You can already summarize a column with `describe()`. So what does a picture add
that a table of statistics does not?

The honest answer is a demonstration rather than an argument, and it is nearly
fifty years old.

Francis Anscombe published four small datasets in 1973 for
exactly this purpose. They are chosen so that the summary statistics a careful
analyst would reach for first are, to two decimal places, the same for all four.
If those statistics were enough, the four datasets would be interchangeable.

````python
# Anscombe's quartet (Anscombe, 1973). Four x/y datasets, eleven points each.
ans = {
    "I":   ([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
            [8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68]),
    "II":  ([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
            [9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74]),
    "III": ([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5],
            [7.46, 6.77, 12.74, 7.11, 7.81, 8.84, 6.08, 5.39, 8.15, 6.42, 5.73]),
    "IV":  ([8, 8, 8, 8, 8, 8, 8, 19, 8, 8, 8],
            [6.58, 5.76, 7.71, 8.84, 8.47, 7.04, 5.25, 12.50, 5.56, 7.91, 6.89]),
}

for name, (x, y) in ans.items():
    x, y = np.array(x), np.array(y)
    slope, intercept = np.polyfit(x, y, 1)
    print(f"{name:4} mean x {x.mean():.2f}  mean y {y.mean():.2f}  "
          f"sd y {y.std(ddof=1):.2f}  r {np.corrcoef(x, y)[0, 1]:.2f}  "
          f"fit y = {slope:.2f}x + {intercept:.2f}")
````

> **Predict.** Four datasets, one set of summary statistics. Sketch what you think the four scatter plots look like.
>
> Commit to an answer before you run the next cell.

````python
fig, axes = plt.subplots(2, 2, figsize=(7, 5.5), sharex=True, sharey=True,
                         layout="constrained")

grid = np.array([3, 20])
for ax, (name, (x, y)) in zip(axes.flat, ans.items()):
    x, y = np.array(x), np.array(y)
    slope, intercept = np.polyfit(x, y, 1)
    ax.scatter(x, y, s=28)
    ax.plot(grid, slope * grid + intercept, linewidth=1.2)   # the same line, four times
    ax.set_title(f"Dataset {name}")

fig.suptitle("Same means, same spread, same correlation, same fitted line")
fig.supxlabel("x")
fig.supylabel("y")

plt.show()
````

**What happened.** One is the relationship the statistics describe. Two is
curved, so a straight line is the wrong summary of it. Three is a straight line
plus a single point pulling the fit upward. Four has no relationship at all: one
point at x = 19 manufactures the entire correlation. The statistics could not
tell you any of that, and one glance can.

## Three things a figure is for

From the course lecture, and worth keeping in this order, because it is also the
order the capstone runs in:

| Purpose | Audience | The question it answers |
|---|---|---|
| **Explore** | you | what is in here that I did not expect? |
| **Explain** | a colleague | here is what I found, and why I believe it |
| **Persuade** | the Cabinet | here is what we should do about it |

The same numbers can serve all three, and the figure that serves one badly
serves the others worse. An exploratory chart is allowed to be ugly and fast. A
chart in a Cabinet report is not.

## Which chart answers which question

Choosing a chart type by how it looks, rather than by the structure of the data
and the question being asked, is how a figure ends up unreadable. Each type has a
job:

| Chart | Use it when | What it shows | Built in |
|---|---|---|---|
| **Histogram** | one continuous variable | the shape of a distribution: skew, spread, whether there is one hump or two | section 3 |
| **Box plot** | comparing distributions across categories | median, quartiles and outliers, compactly enough to line several up | sections 3 and 5 |
| **Scatter plot** | two continuous variables | relationship, clusters and outliers. Color or size can carry a third variable | sections 4 and 6 |
| **Bar chart** | comparing discrete categories | one value per category, compared by length. Start the axis at zero | section 5 |
| **Line chart** | change over an ordered variable, usually time | trend and continuity. The line claims the values between the points exist | section 9 |
| **Heatmap** | a matrix of values | which variables move together, most often a correlation matrix | section 11 |

Three to be wary of, all of which section 9 takes apart: **pie charts**, because
comparing angles is harder than comparing lengths; **three-dimensional charts**,
where the depth carries no data and distorts the comparison; and **dual-axis
charts**, which let the person drawing them decide how related two series
appear.

# 2. The Figure and the Axes

Every figure you make from here on starts with the same line:

```python
fig, ax = plt.subplots()
```

Two objects come back, and knowing which is which explains most of matplotlib.

````python
# alt: An empty figure with a single set of axes, showing the two objects.
fig, ax = plt.subplots(figsize=(5, 3), layout="constrained")

print(type(fig))
print(type(ax))

plt.show()
````

| Object | What it is | What it owns |
|---|---|---|
| **Figure** | the whole image | its size, the title over everything, saving to a file |
| **Axes** | one plot inside that image | the data, the x and y axis, the labels, the legend |
| **Axis** | a single scale on an Axes | its ticks, its limits, its formatting |
| **Artist** | anything drawn | every line, dot, label and spine is one |

An `Axes` means a single set of axes, which is one panel of the figure, while an
axis is one scale on it. The two words look like a singular and a plural of each other, which is
what makes the naming confusing the first time and clear after that.

The practical rule: **if it is about the picture, ask the Figure. If it is about
the data, ask the Axes.**

````python
fig, ax = plt.subplots(figsize=(6, 3.5), layout="constrained")

ax.scatter(cohort.hs_gpa, cohort.ug_gpa, s=26, alpha=0.75)

ax.set_xlabel("high school GPA")          # the Axes owns the labels
ax.set_ylabel("undergraduate GPA")
ax.set_title("One Axes, one claim")
fig.suptitle("The Figure owns this line", fontsize=10)  # and the Figure the frame

plt.show()
````

## The other interface, once

matplotlib has a second way to draw the same figure, and you will meet it
constantly, so it is worth seeing exactly once.

The **implicit** interface, also called the pyplot interface, keeps track of the
last figure and axes you touched and adds things to whatever it thinks you meant.
The **explicit** interface, the one this module uses, hands you the objects and
makes you name them.

Here is one figure, drawn both ways.

````python
# implicit: no objects are named, and there is no way to refer back
plt.plot([1, 2, 3, 4])
plt.ylabel("some numbers")

plt.show()
````

````python
# explicit: the same figure, with the objects named
fig, ax = plt.subplots()
ax.plot([1, 2, 3, 4])
ax.set_ylabel("some numbers")

plt.show()
````

Identical output, four lines each. On one plot the difference is
preference. Here is where it stops being preference.

> **Predict.** You have three series with very different magnitudes: about 1 to 4, about 100 to 400, and about 1000 to 4000. If you call `plt.plot()` three times in a row, what do you get?
>
> Commit to an answer before you run the next cell.

````python
plt.plot([1, 2, 3, 4])
plt.plot([100, 200, 300, 400])
plt.plot([1000, 2000, 3000, 4000])
plt.ylabel("some numbers")

plt.show()
````

**What happened.** All three landed on the same Axes, because that is what the
implicit interface is for: it adds to the current one. The first series is now a
flat line at the bottom, unreadable, because it shares a y axis with values a
thousand times larger. Nothing failed. You just cannot see two thirds of your
data.

````python
# The explicit fix: three Axes, named, each with its own scale.
fig, axes = plt.subplots(3, 1, figsize=(5.5, 5), layout="constrained")

series = [[1, 2, 3, 4], [100, 200, 300, 400], [1000, 2000, 3000, 4000]]
for ax, values in zip(axes, series):
    ax.plot(values)
    ax.set_ylabel("numbers")

axes[-1].set_xlabel("index")
fig.suptitle("Three scales, three Axes")

plt.show()
````

## Translating between them

You will need to read the implicit form, because most examples online use it.
The mapping is nearly mechanical:

| Implicit | Explicit |
|---|---|
| `plt.plot(x, y)` | `ax.plot(x, y)` |
| `plt.xlabel("t")` | `ax.set_xlabel("t")` |
| `plt.title("...")` | `ax.set_title("...")` |
| `plt.xlim(0, 10)` | `ax.set_xlim(0, 10)` |
| `plt.xticks(pos, labels)` | `ax.set_xticks(pos, labels)` |
| `plt.legend()` | `ax.legend()` |
| `plt.savefig("f.png")` | `fig.savefig("f.png")` |

The pattern: a `plt.` function that *sets* something becomes `ax.set_something`.
A `plt.` function that *draws* something keeps its name and moves onto the
Axes.

## Why most examples you find will look different

This is worth knowing before you start searching, because it will otherwise read
as though the course and the internet disagree.

Around 2021 and 2022 the reference material settled on the explicit interface.
matplotlib's own documentation now says:

> In general, we suggest using the OO style, particularly for complicated plots,
> and functions and scripts that are intended to be reused as part of a larger
> project. However, the pyplot style can be very convenient for quick
> interactive work.

The tutorial sites largely did not follow. Roughly two thirds of the widely used
introductions still lead with the implicit interface, and that includes several
whose pages carry a recent update date, because a page can be revised for
formatting without its code being modernized. One well known reference site
teaches plotting without ever using the words Figure or Axes.

So expect a mismatch, and read it as a historical artifact rather than a
disagreement about what is correct. The translation table above is the tool for
it.

Two honest notes on our own rule. First, matplotlib *suggests* the explicit
style, which is weaker than a requirement, and the implicit style is genuinely
convenient for a throwaway look at something. This module uses the explicit form
everywhere anyway, for consistency and because every multi-panel figure needs it.
Second, `plt.subplots()` is itself a pyplot function. "Explicit" does not mean
avoiding `plt.`; it means that once the Figure and Axes exist, you address them
by name.

You will also see `plt.show()` at the end of every figure in this module. A
notebook draws the figure when the cell ends, so there it changes nothing. A
script does not, and leaves you with no window at all. It is here because the
habit transfers to the place where it matters.

**Takeaway.** `fig, ax = plt.subplots()` is the first line of every figure in
this module. The Figure owns the image, the Axes owns one plot inside it. The
implicit interface adds to whatever axes it thinks you meant, which is convenient
until there is more than one, and there is nearly always more than one.

# 3. One variable: what does the spread look like?

The first question about any column is what its values do: where they sit, how
far they range, whether they cluster in one place or two. The histogram answers
it by counting values into bins and drawing the counts.

That word *bins* is doing more work than it looks.

````python
scores = cohort.course_score
print(f"n = {scores.notna().sum()}, missing = {scores.isna().sum()}")

fig, ax = plt.subplots(figsize=(6, 3.2), layout="constrained")
ax.hist(scores, bins=12, edgecolor="white")
ax.set_xlabel("course score")
ax.set_ylabel("students")
ax.set_title(f"Course scores, {scores.notna().sum()} students, 12 bins")

plt.show()
````

> **Predict.** The same 100 scores, drawn with 5 bins and then with 25. Does the shape you just saw survive both?
>
> Commit to an answer before you run the next cell.

````python
fig, axes = plt.subplots(1, 3, figsize=(10, 3.2), sharey=True,
                         layout="constrained")

for ax, n_bins in zip(axes, [5, 12, 25]):
    counts, _, _ = ax.hist(scores, bins=n_bins, edgecolor="white")
    ax.set_xlabel("course score")
    ax.set_title(f"{n_bins} bins (tallest bar: {int(counts.max())})")

axes[0].set_ylabel("students")
fig.suptitle("One column, three bin counts, three impressions")

plt.show()
````

**What happened.** At 5 bins the tallest bar holds 37 of the 100 students, and
the distribution looks like one solid block. At 25 bins the tallest holds 10, one
bin is empty, and the same data looks ragged and possibly bumpy. Nothing about
the students changed between those three pictures. The bin width is a choice you
made, and it is a choice your reader cannot see.

## Choosing a bin count honestly

There is no correct number, which is exactly why it has to be a deliberate
choice rather than a default you never looked at.

- **Too few** bins and real structure disappears into one block.
- **Too many** and sampling noise looks like structure.
- **Look at two or three** before you settle, which is what you just did.

matplotlib will also compute a bin count for you from the data, using rules that
account for both the spread and the sample size.

````python
for rule in ["auto", "sturges", "fd"]:
    edges = np.histogram_bin_edges(scores.dropna(), bins=rule)
    print(f"{rule:8} -> {len(edges) - 1} bins, width {edges[1] - edges[0]:.1f}")
````

**Takeaway.** Report the bin count, or use a named rule and say which. A
histogram without that information cannot be reproduced, and the reader has no
way to know whether the shape is in the data or in your choice.

## The boxplot: the same column, compressed

A boxplot throws away the shape and keeps the summary: median, the middle half,
the reach of the whiskers, and anything beyond them drawn as a point.

````python
fig, axes = plt.subplots(1, 2, figsize=(9, 3.2), layout="constrained")

axes[0].hist(scores, bins=12, edgecolor="white")
axes[0].set_xlabel("course score")
axes[0].set_ylabel("students")
axes[0].set_title("Histogram: shape")

axes[1].boxplot(scores.dropna(), vert=False, tick_labels=["all students"])
axes[1].set_xlabel("course score")
axes[1].set_title("Boxplot: summary")

fig.suptitle(f"Median {scores.median():.1f}, "
             f"middle half {scores.quantile(.25):.1f} to {scores.quantile(.75):.1f}")

plt.show()
````

Note `tick_labels=` in that call. It used to be `labels=`, which was

## Error bars, and saying what they are

An error bar is the usual fix for a bar chart that hides its spread. It is only
useful if the reader knows what the bar represents, because a standard deviation
and a standard error of the mean differ by a factor of the square root of n and
look identical on the page.

````python
sd = by_group["std"]
sem = by_group["std"] / np.sqrt(by_group["count"])

fig, axes = plt.subplots(1, 2, figsize=(9, 3.5), sharey=True, layout="constrained")

for ax, err, name in [(axes[0], sd, "standard deviation"),
                      (axes[1], sem, "standard error of the mean")]:
    ax.bar([0, 1], means, width=0.6, yerr=err, capsize=6)
    ax.set_xticks([0, 1], labels)
    ax.set_title(f"error bars: {name}")

axes[0].set_ylabel("course score")
fig.suptitle("Same data, same bars, error bars that mean different things")

plt.show()
````

**Takeaway.** Always label what the error bar is. Standard deviation
describes the students; standard error describes how well the mean is pinned
down, and it shrinks as the sample grows. A reader who assumes the wrong one will
read the wrong amount of certainty off your figure.

## Category order

Categories with no natural order should be sorted by value, because a reader
compares neighboring bars first. Alphabetical order is a sorting of the labels
rather than the data, and it makes the reader do the work.

````python
counts = cohort.hs_clubs.value_counts()

fig, axes = plt.subplots(1, 2, figsize=(9, 3.2), sharey=True, layout="constrained")

alpha_order = counts.sort_index()
axes[0].bar(alpha_order.index.astype(str), alpha_order.values)
axes[0].set_title("By club count")
axes[0].set_ylabel("students")

by_size = counts.sort_values(ascending=False)
axes[1].bar(by_size.index.astype(str), by_size.values)
axes[1].set_title("By frequency")

fig.suptitle("Order is a choice, and it decides what the reader compares first")

plt.show()
````

**What happened.** When the category is itself a number, as club count is, the
natural order is the numeric one and sorting by frequency destroys information.
When the category is a name with no inherent order, sorting by value is nearly
always the better figure. So the rule to carry is "know which kind of category
you have", which decides the order for you.

````python
sub = sub.assign(band=pd.qcut(sub.hs_gpa, 3,
                              labels=["lower third", "middle third", "upper third"]))

fig, axes = plt.subplots(1, 3, figsize=(11, 3.6), sharex=True, sharey=True,
                         layout="constrained")

for ax, (name, g) in zip(axes, sub.groupby("band", observed=True)):
    r = g.hs_clubs.corr(g.course_score)
    ax.scatter(g.hs_clubs, g.course_score, s=26, alpha=0.75)
    ax.set_xlabel("high school clubs")
    ax.set_title(f"{name}\nGPA {g.hs_gpa.min():.2f} to {g.hs_gpa.max():.2f}\n"
                 f"r = {r:.2f}, n = {len(g)}")

axes[0].set_ylabel("course score")
fig.suptitle("Inside a GPA band, clubs and score barely move together")

plt.show()
````

**What happened.** Across the whole cohort the correlation is 0.43. Inside the

essentially disappeared, which means it was never between clubs and scores. It
was between GPA and both of them: stronger students join more clubs, and stronger
students score better.

````python
# The same result as one number, for the report: the correlation
# between clubs and score after removing what GPA explains of each.
resid_clubs = sub.hs_clubs - np.polyval(np.polyfit(sub.hs_gpa, sub.hs_clubs, 1), sub.hs_gpa)
resid_score = sub.course_score - np.polyval(np.polyfit(sub.hs_gpa, sub.course_score, 1), sub.hs_gpa)

print(f"raw correlation                : {r_raw:+.2f}")
print(f"after controlling for hs_gpa   : {np.corrcoef(resid_clubs, resid_score)[0, 1]:+.2f}")
````

**Takeaway.** 0.43 becomes -0.03. Funding clubs to raise course scores would

Three things are doing the work there.

`axes.flat` walks the 2 by 2 array as a flat sequence, so the loop does not need
to know the grid shape. `sharey=True` gives every panel the same y axis, which is
what makes the four comparable at a glance. And `layout="constrained"` sizes the
gaps so the labels do not collide.

That last one is worth a note, because you will see the older tool everywhere.

## A word on layout

matplotlib's original layout helper was `tight_layout()`, called after the
figure was built. The current documentation describes it as superseded and points
to constrained layout instead, which you request when the figure is created:

```python
fig, axes = plt.subplots(2, 2, layout="constrained")
```

Constrained layout handles the cases `tight_layout` struggles with, including
colorbars and panels that span rows. One trap worth knowing: **calling
`tight_layout()` on a figure turns constrained layout off**, so mixing the two
leaves you with the older behavior and no warning.

> **Predict.** The same four panels without `sharey`. What changes about how the four relationships compare?
>
> Commit to an answer before you run the next cell.

````python
fig, axes = plt.subplots(2, 2, figsize=(9, 6.5), layout="constrained")

for ax, (col, label) in zip(axes.flat, predictors):
    sub = cohort[[col, "course_score"]].dropna()
    ax.scatter(sub[col], sub.course_score, s=24, alpha=0.7)
    ax.set_xlabel(label)
    ax.set_ylabel("course score")

fig.suptitle("The same four panels, each on its own y axis")

plt.show()
````

**What happened.** Every panel now fills its own space, so all four look
similarly spread out and similarly steep. The differences between the four
relationships, which is the entire reason for putting them side by side, are gone.
In a comparison figure the shared axis is the thing that makes it a comparison,
which puts it well above a formatting preference.

## When the grid is not a grid

`plt.subplots` gives you even rows and columns. When the report needs one wide
panel above two narrow ones, `plt.subplot_mosaic` lets you draw the layout and
name the panels, which is easier to read six months later than `axes[1, 0]`.

````python
fig, axd = plt.subplot_mosaic([["overview", "overview"],
                               ["engagement", "prior"]],
                              figsize=(9, 6), layout="constrained")

axd["overview"].hist(cohort.course_score, bins=12, edgecolor="white")
axd["overview"].set_xlabel("course score")
axd["overview"].set_ylabel("students")
axd["overview"].set_title("The outcome we are explaining")

for key, col, label in [("engagement", "logins_per_week", "logins per week"),
                        ("prior", "hs_gpa", "high school GPA")]:
    sub = cohort[[col, "course_score"]].dropna()
    axd[key].scatter(sub[col], sub.course_score, s=22, alpha=0.7)
    axd[key].set_xlabel(label)
    axd[key].set_ylabel("course score")
    axd[key].set_title(f"r = {sub[col].corr(sub.course_score):.2f}")

fig.suptitle("One figure, three panels, named rather than numbered")

plt.show()
````

**Takeaway.** `plt.subplots(rows, cols)` for an even grid, `axes.flat` to
loop it, `sharey=True` whenever the panels are meant to be compared, and
`plt.subplot_mosaic` when the layout is irregular. Requesting
`layout="constrained"` at creation is the current way to keep labels from
colliding.

# 9. What makes a graph bad

Three failures, and they are worth keeping separate because they call for
different fixes:

| Failure | What it means |
|---|---|
| **Bad design** | ugly, cluttered, confusing |
| **Bad data** | cherry-picked or misrepresented |
| **Bad perception** | hard to interpret |

Every example below uses the cohort you have been working with, and every pair
plots **the same numbers twice.** None of the misleading versions required a
single false value, which is the part worth remembering.

## Bad design

The data is fine and the chart is hard to read anyway.

````python
means = cohort.groupby("success").course_score.mean()
labels = ["did not succeed", "succeeded"]

fig, axes = plt.subplots(1, 2, figsize=(10, 4), layout="constrained")

# Cluttered: a legend for two bars that are already labeled, gridlines drawn on
# top of the data, a rotated axis, and a hatch pattern carrying no information.
axes[0].bar([0, 1], means, width=0.6, hatch="///", edgecolor="black",
            label="mean score")
axes[0].grid(True, zorder=5, linewidth=1.4)
axes[0].set_xticks([0, 1], labels, rotation=45)
axes[0].legend(loc="upper left")
axes[0].set_ylabel("mean course score")
axes[0].set_title("Mean Course Score By Success Group (Bar Chart)")

# The same two numbers.
axes[1].bar([0, 1], means, width=0.6)
axes[1].set_xticks([0, 1], labels)
axes[1].set_ylabel("mean course score")
axes[1].set_title(f"Successful students average "
                  f"{means.iloc[1] - means.iloc[0]:.0f} points higher")

plt.show()
````

**What happened.** Every number in the left panel is accurate. It spends the
reader's attention on a hatch pattern, a legend that names what the y axis already names,
gridlines competing with the bars, and a title that describes the chart type. The
right panel spends it on the finding.

## Bad data

Now the harder category. Nothing here is fabricated. Each figure simply chooses
what to show.

> **Predict.** The two group means are 58.7 and 74.8. If the y axis started at 55 instead of 0, how much larger would the gap look?
>
> Commit to an answer before you run the next cell.

````python
fig, axes = plt.subplots(1, 2, figsize=(9, 4), layout="constrained")

for ax, bottom, note in [(axes[0], 55, "y axis starts at 55"),
                         (axes[1], 0, "y axis starts at 0")]:
    ax.bar([0, 1], means, width=0.6)
    ax.set_ylim(bottom, 80)
    ax.set_xticks([0, 1], labels)
    ax.set_title(note)

axes[0].set_ylabel("mean course score")
ratio_truncated = (means.iloc[1] - 55) / (means.iloc[0] - 55)
ratio_honest = means.iloc[1] / means.iloc[0]
fig.suptitle(f"Truncated, one bar looks {ratio_truncated:.1f}x the other. "
             f"From zero, it is {ratio_honest:.2f}x")

plt.show()
````

**What happened.** Truncating the axis turns a ratio of 1.27 into an apparent
ratio of 5.4. Both charts plot 58.7 and 74.8. For a bar chart, where the reader

misrepresents the comparison. A line chart of a quantity that never approaches

````python
# The dual axis: two series, two scales, one implied relationship.
band = cohort.assign(band=pd.qcut(cohort.hs_gpa, 3, labels=["lower", "middle", "upper"]))
by_band = band.groupby("band", observed=True).agg(
    score=("course_score", "mean"), logins=("logins_per_week", "mean"))

fig, axes = plt.subplots(1, 2, figsize=(10, 4), layout="constrained")

ax1 = axes[0]
ax2 = ax1.twinx()
ax1.plot(range(3), by_band.score, marker="o", color="tab:blue")
ax2.plot(range(3), by_band.logins, marker="s", color="tab:red")
ax1.set_ylabel("mean course score", color="tab:blue")
ax2.set_ylabel("mean logins per week", color="tab:red")
ax1.set_xticks(range(3), by_band.index)
ax1.set_title("Two y axes: the lines look locked together")

axes[1].plot(range(3), by_band.score / by_band.score.iloc[0], marker="o",
             label="course score")
axes[1].plot(range(3), by_band.logins / by_band.logins.iloc[0], marker="s",
             label="logins per week")
axes[1].set_xticks(range(3), by_band.index)
axes[1].set_ylabel("relative to the lower band")
axes[1].legend()
axes[1].set_title("One axis: they move at very different rates")

plt.show()
````

## Bad perception

The data is honestly represented and the reader's eye still gets it wrong.

````python
# Area against length. Both panels encode the same two means.
fig, axes = plt.subplots(1, 2, figsize=(9, 4), layout="constrained")

for i, (m, label) in enumerate(zip(means, labels)):
    axes[0].scatter([i], [1], s=(m ** 2) / 4, alpha=0.75)
    axes[0].annotate(f"{m:.1f}", (i, 1), ha="center", va="center")
axes[0].set_xlim(-0.6, 1.6)
axes[0].set_xticks([0, 1], labels)
axes[0].set_yticks([])
axes[0].set_title("Encoded as area")

axes[1].bar([0, 1], means, width=0.6)
axes[1].set_xticks([0, 1], labels)
axes[1].set_ylabel("mean course score")
axes[1].set_title("Encoded as length")

fig.suptitle("A 27 percent difference, read two ways")

plt.show()
````

**What happened.** People compare lengths accurately and areas poorly, and they
tend to underestimate an area difference. Encoding a quantity as the area of a
circle asks the reader to do the harder task for no gain. The same objection
applies to a chart drawn in three dimensions when the third dimension carries no
data: the depth is decoration that distorts the comparison.

````python
# Color implying a grouping the data does not support.
split = cohort.assign(
    area=np.where(cohort.hs_pop_density > cohort.hs_pop_density.median(),
                  "denser area", "less dense area"))
gap = (split.groupby("area").course_score.mean().diff().abs().iloc[-1])

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True, layout="constrained")

for name, g in split.groupby("area"):
    axes[0].scatter(g.hs_gpa, g.course_score, s=28, alpha=0.8, label=name)
axes[0].legend()
axes[0].set_xlabel("high school GPA")
axes[0].set_ylabel("course score")
axes[0].set_title("Colored by a median split")

axes[1].scatter(cohort.hs_gpa, cohort.course_score, s=28, alpha=0.7, color="grey")
axes[1].set_xlabel("high school GPA")
axes[1].set_title(f"Uncolored: the group means differ by {gap:.1f} points")

plt.show()
````

**What happened.** Color reads as category, and a category reads as a real
difference between two kinds of student. The two groups here differ by about 3

This is also the ethics point from the capstone in visual form: population

**Takeaway.** The three checks, in the order they are worth running:

**Design.** Is anything on this figure not helping it say its sentence?

**Data.** Does the axis start where the comparison requires? Is this every
observation, or a filtered subset? Did I choose the bins, the bands, or the split
point after seeing what they did to the picture?

**Perception.** Am I asking the reader to compare lengths, or something harder?
Does color mean a category that is really there?

A figure that passes all three can still be wrong. A figure that fails one of
them is wrong in a way the reader has no way to detect, which is worse.

# 10. Color that other people can read

Color is the fastest channel a figure has and the easiest one to misuse. Two
questions decide almost every color choice: what kind of variable is being
encoded, and who has to read the result.

## Four kinds of colormap, four kinds of question

matplotlib groups its colormaps into four classes, and the class is chosen by the
data rather than by taste:

| Class | Use it when | Examples |
|---|---|---|
| **Sequential** | the value has an order and one end matters | `viridis`, `plasma`, `Blues` |
| **Diverging** | there is a meaningful middle, and both directions matter | `coolwarm`, `RdBu`, `PuOr` |
| **Cyclic** | the value wraps around, like an angle or time of day | `twilight`, `hsv` |
| **Qualitative** | the categories have no order at all | `tab10`, `Set2`, `petroff10` |

The mistake this taxonomy prevents is using a qualitative palette for an ordered
variable, which tells the reader the categories are unrelated when they are a
sequence.

````python
# Sequential: one ordered variable, colored by how much of it there is.
sub = cohort[["hs_gpa", "course_score", "logins_per_week"]].dropna()

fig, ax = plt.subplots(figsize=(6.5, 4.2), layout="constrained")
points = ax.scatter(sub.hs_gpa, sub.course_score, c=sub.logins_per_week,
                    cmap="viridis", s=40)
fig.colorbar(points, ax=ax, label="logins per week")

ax.set_xlabel("high school GPA")
ax.set_ylabel("course score")
ax.set_title(f"Prior record and engagement, together (n = {len(sub)})")

plt.show()
````

`fig.colorbar(points, ax=ax, ...)` asks the **Figure** for the
colorbar and tells it which Axes to steal space from, which is the explicit form
of what `plt.colorbar()` guesses.

`viridis` is matplotlib's default for a reason worth knowing: it is
*perceptually uniform*, meaning equal steps in the data look like equal steps in
color. The older `jet` and `rainbow` maps are not, so they invent boundaries where
the data has none, and both go light in the middle and dark at both ends, which
makes them unreadable in grayscale.

````python
# Diverging: distance from a meaningful middle, in two directions.
mean_score = cohort.course_score.mean()
deviation = cohort.course_score - mean_score
limit = deviation.abs().max()

fig, ax = plt.subplots(figsize=(6.5, 4.2), layout="constrained")
points = ax.scatter(cohort.hs_gpa, cohort.hs_attendance, c=deviation,
                    cmap="coolwarm", vmin=-limit, vmax=limit, s=40)
fig.colorbar(points, ax=ax, label=f"points above or below the mean of {mean_score:.1f}")

ax.set_xlabel("high school GPA")
ax.set_ylabel("high school attendance")
ax.set_title("A diverging map has to be centered, or the middle color lies")

plt.show()
````

**Takeaway.** `vmin=-limit, vmax=limit` is what makes that figure honest. A
diverging colormap puts its neutral color at the midpoint of the range it is
given, so if the range is not symmetric around zero, the color that reads as "at
the mean" sits somewhere else. This is the color version of a truncated axis.

## Palettes that survive a colorblind reader, and a printer

About one man in twelve has some form of color vision deficiency, most often
affecting red and green. matplotlib's default categorical cycle, `tab10`, is not
documented as safe for that, and recent versions added sequences that are.

````python
import matplotlib

# These arrived in different matplotlib versions, so ask rather than assume.
preferred = ["okabe_ito", "petroff6", "petroff10", "tab10"]
available = [name for name in preferred if name in matplotlib.color_sequences]

print(f"matplotlib {matplotlib.__version__}")
print("available from the preferred list:", available)
print("using:", available[0])

palette = matplotlib.color_sequences[available[0]]
````

`okabe_ito` and `petroff6` were added in matplotlib 3.11, and
`petroff10` in 3.10. If your version does not have them, the cell above falls
back and the figures still work. Checking `matplotlib.color_sequences` rather
than hardcoding a name is the habit worth taking away: the library moves, and
code that asks what is present outlives code that assumes.

````python
bands = cohort.assign(band=pd.qcut(cohort.hs_gpa, 3,
                                   labels=["lower third", "middle third", "upper third"]))

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True, layout="constrained")

# Color alone.
for (name, g), c in zip(bands.groupby("band", observed=True), palette):
    axes[0].scatter(g.logins_per_week, g.course_score, s=34, alpha=0.85,
                    color=c, label=name)
axes[0].legend(title="high school GPA")
axes[0].set_xlabel("logins per week")
axes[0].set_ylabel("course score")
axes[0].set_title("Encoded by color only")

# Color plus a second channel.
for (name, g), c, marker in zip(bands.groupby("band", observed=True), palette,
                                ["o", "s", "^"]):
    axes[1].scatter(g.logins_per_week, g.course_score, s=34, alpha=0.85,
                    color=c, marker=marker, label=name)
axes[1].legend(title="high school GPA")
axes[1].set_xlabel("logins per week")
axes[1].set_title("Encoded by color and shape")

plt.show()
````

**Takeaway.** Never let color be the only thing carrying a distinction. A
second channel, marker shape for points, line style for lines, direct labels for
bars, costs one argument and makes the figure work for a reader who cannot
separate two of your colors, and for anyone who prints it.

> **Predict.** Take the palette you just used and convert each color to the gray it would print as. Do the categories stay distinguishable?
>
> Commit to an answer before you run the next cell.

````python
# alt: Two strips of color swatches, the palette in color above and the
# alt: same colors converted to the gray they would print as below.
import matplotlib.colors as mcolors

fig, axes = plt.subplots(2, 1, figsize=(7, 2.4), layout="constrained")
show = palette[:6]

for i, c in enumerate(show):
    axes[0].add_patch(plt.Rectangle((i, 0), 0.9, 1, color=c))
    # Relative luminance, the same quantity a contrast check uses.
    r, g, b = mcolors.to_rgb(c)
    gray = 0.2126 * r + 0.7152 * g + 0.0722 * b
    axes[1].add_patch(plt.Rectangle((i, 0), 0.9, 1, color=(gray, gray, gray)))
    axes[1].text(i + 0.45, 0.5, f"{gray:.2f}", ha="center", va="center",
                 color="white" if gray < 0.5 else "black")

for ax, title in zip(axes, [f"{available[0]} in color", "the same colors as printed gray"]):
    ax.set_xlim(0, len(show))
    ax.set_ylim(0, 1)
    ax.set_xticks([])
    ax.set_yticks([])
    ax.set_title(title, fontsize=10)

plt.show()
````

**What happened.** Two colors with similar luminance become the same gray, and no
amount of hue difference rescues them once the figure is printed or photocopied.
The numbers printed on the lower strip are relative luminance on a 0 to 1 scale;
categories whose values sit close together need a second channel or a different
palette.

# 11. Two shortcuts: pandas and seaborn

Everything so far has been matplotlib. Two libraries sit on top of it and will
draw a figure in one line, and both hand you back matplotlib objects, so
everything from sections 2 and 8 still applies.

## pandas: `.plot()` on the frame you already have

`DataFrame.plot()` and `Series.plot()` accept a `kind` and, importantly, an `ax`.
Passing the Axes you made keeps the figure yours.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 3.6), layout="constrained")

# The matplotlib version.
axes[0].hist(cohort.course_score, bins=12, edgecolor="white")
axes[0].set_xlabel("course score")
axes[0].set_ylabel("students")
axes[0].set_title("matplotlib")

# The pandas version, drawn onto an Axes we made.
cohort.course_score.plot(kind="hist", bins=12, edgecolor="white", ax=axes[1])
axes[1].set_xlabel("course score")
axes[1].set_title("pandas, with ax=")

fig.suptitle("Same figure, and the second one is one line")

plt.show()
````

````python
# .plot() returns the Axes, which is why the rest still works.
returned = cohort.course_score.plot(kind="hist", bins=12)
print(type(returned))
returned.set_title("Still an Axes, so set_title is still the method")
````

**Takeaway.** Always pass `ax=`. Without it, pandas draws on whatever Axes
matplotlib thinks is current, which is the implicit interface arriving through a
side door. With it, `.plot()` is a convenient way to fill an Axes you control.

`kind=` covers `line`, `bar`, `barh`, `hist`, `box`, `kde`, `area`, `scatter` and
`hexbin`. When you need anything the kind does not offer, you already have the
Axes and can keep going in matplotlib.

## seaborn: statistical figures in one call

seaborn's value is the figures that would take a dozen matplotlib lines: a
correlation heatmap, a regression plot with a confidence band, a grid of every
pair of variables.

It divides into two kinds of function, and the difference decides how you use
them.

````python
import seaborn as sns

sns.set_theme(style="whitegrid")   # the current call; sns.set() is the old name

corr = cohort[["hs_gpa", "hs_attendance", "logins_per_week",
               "submission_rate", "course_score"]].corr()

fig, ax = plt.subplots(figsize=(6, 4.8), layout="constrained")
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1,
            ax=ax)
ax.set_title("Correlations, with the diverging scale centered on zero")

plt.show()
````

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharey=True,
                         layout="constrained")

sns.regplot(data=cohort, x="logins_per_week", y="course_score", ax=axes[0])
axes[0].set_title("regplot: fit and confidence band")

sns.boxplot(data=cohort, x="success", y="course_score", ax=axes[1])
axes[1].set_xticks([0, 1], ["did not succeed", "succeeded"])
axes[1].set_xlabel("")
axes[1].set_title("boxplot: groups without the loop")

plt.show()
````

Every seaborn call above passed `ax=`. Those are seaborn's
**axes-level** functions, and `ax=` is how seaborn's own documentation says to
use them with the object-oriented interface.

The other kind is different, and it is worth being explicit about rather than
surprised by.

````python
# A figure-level function builds its OWN figure. It does not take ax=.
grid = sns.pairplot(cohort[["hs_gpa", "logins_per_week", "course_score"]],
                    height=2.0)
grid.figure.suptitle("pairplot returns a grid object, not an Axes", y=1.02)
print(type(grid))

plt.show()
````

**Takeaway.** **The rule for this module.** Axes-level functions
(`scatterplot`, `histplot`, `boxplot`, `regplot`, `heatmap`, `kdeplot`) are
artists: make the Figure yourself and pass `ax=`.

Figure-level functions (`relplot`, `displot`, `catplot`, `lmplot`, `jointplot`,
`pairplot`) are figure constructors: they own their figure, they cannot take
`ax=`, and you customize them through what they return, using `grid.figure`,
`grid.axes`, or `grid.set(...)`.

If a figure has to sit inside a layout you built, that decides it: use the
axes-level function.

# 12. Saving a figure for the report

A figure that looks right in a notebook can be unreadable in a document, and the
reason is almost always the same: the file was written at the wrong resolution,
or in the wrong format.

````python
import os

# Write into a folder rather than beside the notebook. A report generates a
# lot of these, and a working directory with thirty loose PNGs in it is its
# own small problem.
os.makedirs("figures", exist_ok=True)

fig, ax = plt.subplots(figsize=(6, 4), layout="constrained")
ax.scatter(cohort.hs_gpa, cohort.course_score, s=26, alpha=0.7)
ax.set_xlabel("high school GPA")
ax.set_ylabel("course score")
ax.set_title("The figure we are about to save five ways")

for name, kwargs in [("fig_72.png", {"dpi": 72}),
                     ("fig_150.png", {"dpi": 150}),
                     ("fig_300.png", {"dpi": 300}),
                     ("fig.pdf", {}),
                     ("fig.svg", {})]:
    path = os.path.join("figures", name)
    fig.savefig(path, **kwargs)
    print(f"{name:12} {os.path.getsize(path) / 1024:6.0f} KB")

plt.show()
````

**What happened.** The three PNGs hold identical information at 9, 28 and 113
kilobytes. The difference is how many pixels the file contains, which decides
whether the axis labels survive being printed or projected. 72 is a screen
thumbnail, 150 is a reasonable document figure, and 300 is print.

The PDF is 9 kilobytes and loses nothing when it is enlarged, because it stores
the instructions to draw the figure rather than a grid of pixels.

## Raster and vector

| | Formats | Scales without blurring | Good for |
|---|---|---|---|
| **Raster** | PNG, JPG | no | screens, slides, anything with a photograph in it |
| **Vector** | PDF, SVG, EPS | yes | documents, print, anything with text in it |

A figure is mostly text and thin lines, which is exactly what vector formats
handle well. Send a PDF to anyone producing a printed report. Use PNG at 150 or
higher when the destination needs an image file, and avoid JPG for figures, since
its compression smears the edges of text and lines.

````python
# bbox_inches="tight" trims the surrounding whitespace, which
# matters when the figure is being dropped into a document.
fig.savefig("figures/fig_padded.png", dpi=150)
fig.savefig("figures/fig_tight.png", dpi=150, bbox_inches="tight")

for name in ["fig_padded.png", "fig_tight.png"]:
    print(f"{name:16} {os.path.getsize('figures/' + name) / 1024:6.0f} KB")
````

**Takeaway.** `fig.savefig(name, dpi=..., bbox_inches="tight")` is the call
worth memorizing, and note it is the **Figure** that saves itself, not `plt`.

Two habits: save from the same cell that builds the figure, so the file cannot
drift from the code that made it, and give the file a name that says what the
figure claims rather than `figure1.png`.

what you would guess. matplotlib accepted all 100 rows, and 85 of them produced a
visible point. The missing values are still in the collection, held as a masked
array, and they are simply never drawn. So the figure shows 85 students, no
warning was issued, and a naive `len()` on the offsets would have told you 100 and
agreed with the false title.

This is the mechanical argument for D30, the rule that a figure's title states its
finding **and the cell computes the number in it**. A computed n cannot be wrong
in this way.

## Quiet failure 3: the column that is not what you think

A column read from a CSV can arrive as text that looks numeric. Plotting it does
not fail. It produces a category for every distinct value.

````python
as_text = cohort.course_score.round().astype(int).astype(str)

fig, axes = plt.subplots(1, 2, figsize=(10, 3.4), layout="constrained")

axes[0].hist(as_text, bins=12)
axes[0].set_title("course_score as text")
axes[0].tick_params(axis="x", labelrotation=90, labelsize=6)

axes[1].hist(cohort.course_score, bins=12, edgecolor="white")
axes[1].set_title("course_score as numbers")

print("dtype of the text version:", as_text.dtype)

plt.show()
````

**What happened.** The left panel is not a distribution. Each distinct string
became its own category, in the order encountered, so the x axis is neither sorted
nor numeric and the shape means nothing. Check `.dtype` before plotting anything
that came out of a file, which is the same habit Module 4 built for
`submission_rate`.

## Quiet failure 4: figures that pile up

Every figure you create stays in memory until it is closed. In a notebook this is
usually harmless. In a loop that writes fifty figures to disk, it is not.

````python
before = len(plt.get_fignums())

for band in ["lower", "middle", "upper"]:
    fig, ax = plt.subplots()
    ax.set_title(band)
    fig.savefig(f"figures/band_{band}.png", dpi=72)
    plt.close(fig)          # the line that matters

print(f"open figures before: {before}, after: {len(plt.get_fignums())}")

plt.show()
````

**Takeaway.** `plt.close(fig)` after saving, whenever figures are being made
in a loop. Without it matplotlib warns after twenty open figures and keeps every
one of them in memory.

The four quiet failures share a shape: **the code ran, so nothing told you.** The
defense is to check the figure against the data rather than against your
expectation. Count the points. Print the dtype. Ask which Axes you are on.
